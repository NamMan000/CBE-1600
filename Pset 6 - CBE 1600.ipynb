<h5> e). Use the VdW EOS (and tablated physical property data, as needed) to calculate the
density of steam at 400 K at 1 atm. What is the error involved in descirbing the density of steam
using the ideal gas EOS relative to the VdW EOS? </h5>
P_c = 220.64 bars (Source: 	Sato, Watanabe, et al., 1991, NIST: https://webbook.nist.gov/cgi/cbook.cgi?ID=C7732185&Mask=4)
T_c = 647 Kelvin (Source: NIST handbook)

```python
import numpy as np
T = 400
P = 1 #atm
R = 0.08206 #L*atm/mol*K
dens1 = P/(R*T) # in moles/L

gperL = dens1*18 #g/L
print(f"E). The density of the stream is {gperL:.3f} g/L via the ideal gas law")

Tc = 647
Pc = 220.64/1.013 #converting from bars to ATM
a = (27/64)*((R*Tc)**2/Pc)
b = (R*Tc/(8*Pc))

coef = [P, -(P*b+R*T), a, -a*b] #in part B we made a cubic equation and can solve for molar volumes (Volume/1mol) given we know all the other coefficient
roots = np.roots(coef)
dens2 = 1/roots*18 # getting vol/mol into moles/Volume * g/mol for water = g/L
print(f"E). The density is {dens2[0]:.3f} g/L via the VanderWaal's Equation of State, with an error of {(dens2[0]-gperL)/gperL * 100:.2f}%") #chose the first root

```

<h5> a). You and your office-mate—an intern from a different university—have been asked to estimate the

I also got 95 cars when very simply using the ideal gas law. This estimate does not account for the fact that the final pressure did not go to 0, but is instead 55, meaning that all the moles did not leave.

Also, ideal gas law did not work for such high pressures and so we know that we need a different estimate to get the true number of moles and cars powered.  </h6> <b>


```python
Vt = np.pi*(0.56/2)**2 * 10.5 * 1000 #m^3 --> L
Pt = (2600)/14.696 + 1 # psi --> atm
Tt = 298 #K


moles = Pt*Vt/(R*Tt) #IDEAL GAS LAWd escribes the number of moles that are discharged from each tube
def carcalc (m):
    print(f"{m*2.016/100/4:.0f} cars") # a quick function for getting from moles in tube to cars; units go from moles --> grams --> kilograms *10 tubes / 4 kg per cars

print("A). The ideal gas law-based quick estimate:")
carcalc(moles)
```

<h5>b).Do the calculation using the SRK equation of state. Instead of using Eqs. 5.3-11 and 5.3-13
for the parameter α, use the following correlation developed specifically for hydrogen23: 1.202*e^-0.3228Tr </h5>

$\textrm P = \frac {RT}{V_m-b} - \frac{aα}{V_m(V_m+b)}$
here we are now refering to molar volumes, eliminating n.
Rearranged to be cubic:

$\textrm PV_m^3 −(Pb+RT)V_m^2 +aαV_m −aαb=0 $

<b>coefficients</b>

$\textrm α = 1.202*e^-0.3228Tr$

$\textrm a = 0.42747* \frac{(R*Tc)^2}{Pc}$

$\textrm b = 0.08664* \frac{(R*Tc)}{Pc}$

Critical Temperature = <b>33.18K</b> (Source:Onnes, Crommelin, et al., 1917, NIST Webbook: https://webbook.nist.gov/cgi/cbook.cgi?ID=C1333740&Mask=4 )

Critical Pressure = 13.00bars/1.013 (bars/atm) = <b> 12.833 atm </b> (Source: Same as above)


```python
#ROOTS
Tc = 33.18 # K Critical Temperature
Pc = 13.00/1.1013 #bars to atm atm Critical Pressure

#SRK
Tr = Tt/(Tc)
aval = 0.42747*((R*Tc)**2/Pc)
alpha = 1.202*np.e**(-0.3228*Tr)
bval = 0.08664*(R*(Tc)/Pc)

coefs597 = [Pt, -(Pt*bval +R*Tt), aval*alpha, -aval*alpha*bval] # are returned with a molar volume (L/mol)
molarvs = np.roots(coefs597)
molarvs = Vt/float(molarvs[0]) #VolumeTank/VolumePerMol = Number of Moles

print("B). The estimate of number of cars powerable using the V_m roots of the Vanderwaal Equation:")
carcalc(molarvs)
```

```python
#ITERATION
def newV(Vguess):
    return(R*Tt)/(Pt)+bval-(aval*alpha/Pt)*((Vguess - bval)/(Vguess*(Vguess+bval)))

Vguess = R*Tt/Pt

def tolerance(Vold,Vnew):
    return (Vnew-Vold)/Vold

tol = 10
while tol > 0.07:
    Vold = Vguess
    Vguess = newV(Vold)
    tol = tolerance(Vold,Vguess)

print("B). The estimate of number of cars powerable using iteration of the Vanderwaal Equation (should get a number similar to above):")
carcalc(Vt/Vguess)

```
