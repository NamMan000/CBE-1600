Force(Newton*meters = kg*m*s<sup>-2</sup>) occurs when an object with mass accelerating or moving. We know that work (Nm or joule) is the amount of energy needed to move 1 kg object with an acceleration of 1m/s^2 for a length of 1 m.

Similarily, Pressure is the amount of force exerted by the atoms/area[Nm<sup>-2</sup> or kg * m<sup>-1</sup>* s<sup>-2</sup>]. Instead of just getting the force on 1 wall (N/m^2) of the container, we get the energy done over the volume of the container. Hence, pressure is like energy density.
