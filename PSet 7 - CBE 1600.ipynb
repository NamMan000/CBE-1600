<h5> What is Raoult’s Law? Under what conditions does it apply? </h5>

Raoult's Law states that the vapor pressure of a particular species is dependent on the mole fraction of that species in the liquid, and is a product of this mole fraction and the saturated vapor pressure at the particular temperature.

P<sub>i</sub> = x<sub>i</sub>*P<sup>sat</sup>(T)

It applies under ideal conditions, meaning that the intermolecular forces between the different components are similar to the those between like components.

dependence of surface coverage as a function of the concentration or pressure of the species in the

<h5> A). Provide a derivation of the Langmuir adsorption isotherm. Define the terms. <h5>

Let's say there is one type of surfactant that is floating around. If it comes into contactant with a surface, it will move from the mobile phase to the bound state.

$\ Surface (S) + Mobile Surfactant (M) <-> Bound State (SM) $

In this model, we are assuming that the surfactant is like a gas. We have a form of a chemical equation in which we can write an equilibrium coefficient.  

$ \ K_e = \frac {[SM]}{[S][M]} $

Now let's introduce something called the fractional coverage: the percent of sites covered with the surfactant. We assume that this is a reversible reaction (it can be covered or uncovered), independent of the status of the sites around it and whether they are covered, and can only be covered once as a monolayer.

We now write the change in this value in the forward direction.

$\ \frac {dθ}{dt} = K_a [N][1-θ][C_s] $

K_a describes the absorption, but, as we can see, the forward reaction depends on the concentration of particles floating around (S), the percent of available site (1-θ), and the number of open sites.

We now write the change in the fractional coverage in the backward direction (desorption), where we don't care about much besides the desorption constant and the amount of site covered and therefore available to be desorbed:

$\ \frac {dθ}{dt} = -K_d [θ][N] $

Set both derivatives of θ with respect to time equal to each other because we are equilibrium:

$\ (1).  K_a [N][1-θ][C_s] = K_d [θ][N] $

$\ (2).  K_a [N][C_s] - K_a [N][C_s][θ] = K_d [θ][N]$

$\ (3).  K_a [N][C_s] = K_a [N][C_s][θ] + K_d [θ][N]$

$\ (4). K_a [N][C_s] = θ(K_a [N][C_s] + K_d[N])$

$\ (5). \frac {K_a[C_s]}{K_a[C_s] +K_d} =  θ $

$\ (6). \frac {K_a[C_s] +K_d}{K_a[C_s]} =  \frac {1}{θ} $

$\ (7). 1 + \frac {1}{K_c} =  \frac {1}{θ} $

$\ (8). \frac {K_c}{K_c} + \frac {1}{K_c} =  \frac {1}{θ} $

$\ (9). \frac {K_c+1}{K_c}  =  \frac {1}{θ} $

$\ (10). \frac {K_c}{K_c+1}  =  θ $

Toluene: same as above

All data from the same compilation source: NIST Webbook

```python
import numpy as np
pressure = 101325/100000 #pascals to bars
def antoine(P, A, B, C):
    Temperature = (B/(A-np.log10(P)))-C
    return Temperature

water = {
    "A": 4.6543,
    "B" : 1435.264,
    "C" : -64.848,
    "Tr": 100
} #Stull 1947

methanol = {
    "A": 5.20409,
    "B" : 1581.341,
    "C" : -33.50,
    "Tr": 64.7

} #Ambrose & Stark - NIST Webbook

eth = {
    "A": 5.24677,
    "B" : 1598.673,
    "C" : -46.424,
    "Tr": 78.37
} #Ambrose & Stark - NIST Webbook

hex = {
    "A":3.99695,
    "B" : 1171.53,
    "C" : -48.784,
    "Tr": 68.7
} #Williamham, Taylor, et al., 1945 - NIST Webbook


tol = {
    "A": 4.07827,
    "B" : 1343.943,
    "C" : -53.773,
    "Tr": 110.6
} #Williamham, Taylor, et al., 1945 - NIST Webbook


bpwater = antoine(pressure, water["A"], water["B"], water["C"]) -273.15
difw = np.abs(bpwater- water["Tr"])/ water["Tr"]
bpmethanol = antoine(pressure, methanol["A"], methanol["B"], methanol["C"])-273.15
difm = np.abs(bpmethanol - methanol["Tr"])/ methanol["Tr"]
bpethanol = antoine(pressure, eth["A"], eth["B"], eth["C"]) -273.15
dife = np.abs(bpethanol - eth["Tr"])/ eth["Tr"]
bphex = antoine(pressure, hex["A"], hex["B"], hex["C"]) -273.15
difh = np.abs(bphex - hex["Tr"])/ hex["Tr"]
bptol = antoine(pressure, tol["A"], tol["B"], tol["C"]) -273.15
diftol = np.abs(bptol- tol["Tr"])/ tol["Tr"]

print(f"Water BP: {bpwater:.3f} °C. The differene from actual is {difw:.3f}")
print(f"Methanol BP: {bpmethanol:.3f} °C. The differene from actual is {difm:.3f}")
print(f"Ethanol BP: {bpethanol:.3f} °C.The differene from actual is {dife:.3f}")
print(f"Hexane BP: {bphex:.3f} °C. The differene from actual is {difh:.3f}")
print(f"Toluene: {bptol:.3f} °C. The differene from actual is {diftol:.3f}")
```

Clapeyron:
$ \frac{dP}{dT} = \frac{ΔH_v}{T(V_g - V_l)} $

It describes the Change in Vapor Pressure with respect to Temperature. This change is proportional to  by the Heat of Vaporization and inversely proportional to the Temperature to the Vapor Pressure.  We make certain assumption like the volume of the gas is much larger than the volume of liquid, that the heat of vaporization is constant, etc.,

Deriving the Clausius Clapeyron Equation:
$ \frac{dP}{dT} = \frac{ΔH_v}{T(V_g)} $
Volume of gas = RT/P

$ \frac{dP}{dT} = \frac{ΔH_v}{T(RT/P)} $

$ \frac{dP}{dT} = \frac{ΔH_v *P}{T^2*R} $

divide both sides by P to get rid of this tern & multiply by T^2

$ \frac{dP/P}{dT}= \frac{ΔH_v *P}{T^2*R} $

$\frac{dP}{d(T)}* \frac{1}{P*T^2}= = \frac{-ΔH_v}{R} $

The antiderivative of 1/P is ln(P). The antiderivative of T^-2 is -T^-1.

$\frac{dlnP}{d(1/T)} = \frac{-ΔH_v}{R} $




Clausius-Clapeyron:
$\frac{dln(P)}{d(1/T)} = \frac{-ΔH_v}{R} $
