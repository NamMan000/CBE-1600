<h3> Question 1: Dimensionless Numbers<h3>
<h5> a. What is the utility of dimensionless numbers? </h5>
It allows us to describe a property of substances as a ratio of two other quantities. It allows us to simplify our characterization of complex beahviors of fluids, down to a simple number rather than a number attached to a string of units, allowing easy comparisons.
<h5> b. Provide 3 examples of dimensionless numbers commonly used in Chemical
Engineering, and describe in what contexts they find use. </h5>
Pechle Number gives the relation between convective/diffusive forces. If you want some liquid that moves quickly (high convective) and doesn't diffuse the particles in it (which is the spreading out of those particles and getting it dissolved), then look for a higher pechle number. Otherwise, if you don't care about moving quickly, then go for a lower pechle number for more diffusion.

Reynold's Number describes the convection/viscosity. If you want laminar (straight) flow, you need a lower Reynold's number which means very little convection or circulation and a high viscosity (which it resists flow and will go slowly in the same direction). Otherwise, turbulent flow is charactereized by a low Reynold's Number.

Damköhler number relates reaction rate (1/s)/diffusive transport (s). Diffusive transport is given by area/D where D is a coefficient given by area/seconds so you get the number of seconds taken for that area. So you end up understanding the input time of a reactant versus the rate of reaction occuring. A context in which this might be useful is the thermal oxide growth on silicon, which is dictated by both the reaction rate as well as the diffusion of oxygen to the surface for continued growth.

<h3>Problem 3.17</h3>
a). Aqueous solutions of the amino acid L-isoleucine (Ile) are prepared by putting 100.0 grams of pure water into each of six flasks and adding different precisely weighed quantities of Ile to each flask. The densities of the solutions at 50 +/- 0:05°C are then measured with a precision densitometer, with the following results:

Plot a calibration curve showing the mass ratio, r, as a function of solution density, ρ, and fit a
straight line to the data to obtain an equation of the form r  aρ  b.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from scipy.optimize import curve_fit

r = np.array([0.0000,0.8821,1.7683,2.6412,3.4093,4.2064])
p=np.array([0.98803,0.98984,0.99148,0.99297,0.99439,0.99580])

plt.ylabel("r (g Ile/100 g H2O)")
plt.xlabel("ρ (g solution/cm^3)")
plt.scatter(p,r)
plt.title("mass ratio (g Ile/100 g H2O) vs density of solution (g solution/cm^3)")
slope, intercept, r_value, p_value, std_err = linregress(p,r)
regression_line = slope * p  + intercept
plt.plot(p,regression_line, linestyle = "--", color = "red", label = f'y = {slope:.3f}x+{intercept:.3f}')
plt.legend()
```

<h5> b). The volumetric flow rate of an aqueous Ile solution at a temperature of 50°C is 150 L/h. The density of a sample of the stream is measured at 50°C and found to be 0.9940 g/cm3. Use the calibration equation to estimate the mass flow rate of Ile in the stream (kg Ile/h). </h5>

```python
pb = 0.9940
rb = (slope * pb)  + intercept # g Ile/100g H2O
volflowrate = 150 #L

cms = volflowrate * 10**3 #cm^3
print(f"{cms} is volume in in one hour")

mass = cms* pb  #grams/hr
print(f"{mass} is mass in g in one hour")

x= mass*(rb/(rb+100))/1000
print(f"{x:.3f} is mass of Ile in kg in one hour")



```

<h5>For each of the given readings, calculate the pressure drop across the orifice, ΔP(mm Hg).</h5>

$\Delta P (mm Hg) = \frac{(1.10-0.791)g}{cm^3}\space \frac{981 cm}{s^2}\space \frac{h(mm)}{}\space \frac{1 cm}{10 mm}\space \frac{1 dyne}{1\space g\cdot cm/s^2}\space \frac{760\space mm Hg}{1.01325\ast 10^6\space dyne/cm^2}= 0.02274\space h(mm)$


```python
import numpy as np
def pressdrop(height,density):
    resultp = (height*10**-3*density*9.8)/133.3
    return resultp
h = np.array([5,10,15,20,25,30])
p = 1.1 * 10**3 - 791 #kg/m^3 =     g/cm^3 *10^6 cm^3/m^3 * 10^-3 kg/g
hdrop = pressdrop(h,p) #kg/m^3 * N/kg * m = N/m^2
print(f"The Pressure Drop in mmHg:{hdrop}")
```

(b) The flow rate through an orifice should be related to the pressure drop across the orifice by the formula
V=K(ΔP)<sup>n</sup>
Verify graphically that the given orifice calibration data are correlated by this relationship, and
determine the values of K and n that best fit the data

$ln\dot V=n\cdot ln(\Delta P)+lnK$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from scipy.optimize import curve_fit

plt.title(r'ln(V) vs. ln($\Delta$P) Across an Orifice')
vol = [62,87,107,123,138,151]

logV = np.log(vol)
logP = np.log(hdrop)


slope, intercept, r_value, p_value, std_err = linregress(logP, logV)
regression_line = slope * logP  + intercept
plt.plot(logP,slope * logP + intercept, linestyle = "--", color = "blue")
plt.xlabel(r'ln($\Delta$P)')
plt.ylabel(r'ln(V)')

print(f'n = {slope:.3f}')
print(f'K = {np.e**intercept:.0f}' )
```

Suppose the orifice meter is mounted in a process line containing acetone and a reading h=23 mm
is obtained. Determine the volumetric, mass, and molar flow rates of acetone in the line

```python
cpressure = pressdrop(23,p)
print(f'pressure in mmHg {cpressure:.3f} at h = 23 mm')

lnVnew = slope * np.log(cpressure) + intercept #getting the new value based on the equation  in ln form
vflowrate = np.e**lnVnew
print(f"Volumetric Flow Rate is {vflowrate:.0f} mL/s") #printing the volumetric flow rate
sgacetone = 0.791
maflowrate = vflowrate * sgacetone
print(f"Mass Flow Rate is {maflowrate:.0f} g/s")
molarflowrate = maflowrate/58.08
print(f"Molar Flow Rate is {molarflowrate:.2f} mol/s")


```
